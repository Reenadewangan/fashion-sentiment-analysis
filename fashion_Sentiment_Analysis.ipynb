```python
import re

def clean_text(text):

    text = re.sub(r'http\S+|www\S+', '', text)

    text = re.sub(r'[^a-zA-Z\s]', '', text)

    text = text.lower()

    text = re.sub(r'\s+', ' ', text).strip()

    return text


df['cleaned_review'] = df['review_text'].apply(clean_text)

df[['review_text', 'cleaned_review']].head(10)
```

```python
import re
import emoji

def clean_text(text):

    text = text.lower()

    # Convert emojis into text descriptions
    text = emoji.demojize(text)

    # Remove URLs
    text = re.sub(r'http\S+|www\S+', '', text)

    # Keep letters, spaces and emoji descriptions
    text = re.sub(r'[^a-zA-Z_:\s]', '', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python

from sklearn.model_selection import train_test_split

X_text = df['processed_review']
y = df['Sentiment']

X_train_text, X_test_text, y_train, y_test = train_test_split(
    X_text,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train_text))
print("Testing samples:", len(X_test_text))
```

```python

from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

lr_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95
    )),

    ('model', LogisticRegression(
        max_iter=1000,
        class_weight='balanced',
        random_state=42
    ))
])

lr_pipeline.fit(X_train_text, y_train)

print("Pipeline training completed!")
```

```python

y_pred_lr = lr_pipeline.predict(X_test_text)

print("Predictions generated!")
```

```python

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy:", accuracy_score(y_test, y_pred_lr))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lr))
```

```python

from sklearn.naive_bayes import MultinomialNB
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer

nb_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95
    )),

    ('model', MultinomialNB(alpha=1.0))
])

# Calculate weights for minority classes
sample_weights = compute_sample_weight(
    class_weight='balanced',
    y=y_train
)

# Train the model
nb_pipeline.fit(
    X_train_text,
    y_train,
    model__sample_weight=sample_weights
)

print("Naive Bayes training completed!")
```

```python

y_pred_nb = nb_pipeline.predict(X_test_text)

print("Predictions generated!")
```

```python

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

print("Accuracy:", accuracy_score(y_test, y_pred_nb))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_nb))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_nb))
```

```python

from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer

svm_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95
    )),

    ('model', LinearSVC(
        class_weight='balanced',
        random_state=42,
        max_iter=2000
    ))
])

# Train the model
svm_pipeline.fit(X_train_text, y_train)

print("Linear SVM training completed!")
```

```python

from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        max_features=5000,
        ngram_range=(1, 2),
        min_df=2,
        max_df=0.95
    )),
    ('model', LogisticRegression(
        max_iter=1000,
        class_weight='balanced',
        random_state=42
    ))
])

param_grid = {
    'model__C': [0.1, 1, 3, 5]
}

grid_search = GridSearchCV(
    pipeline,
    param_grid,
    cv=3,
    scoring='f1_macro',
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train_text, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best CV Macro F1:", grid_search.best_score_)
```

```python

best_model = grid_search.best_estimator_

y_pred_tuned = best_model.predict(X_test_text)

print("Accuracy:", accuracy_score(y_test, y_pred_tuned))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_tuned))
```

```python

from sklearn.linear_model import SGDClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report

loss_functions = ['hinge', 'modified_huber', 'log_loss']

sgd_results = []

for loss in loss_functions:

    model = SGDClassifier(
        loss=loss,
        penalty='l2',
        alpha=0.0001,
        max_iter=2000,
        class_weight='balanced',
        random_state=42,
        tol=1e-3
    )

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)

    macro_f1 = f1_score(
        y_test, predictions, average='macro'
    )

    accuracy = accuracy_score(y_test, predictions)

    sgd_results.append({
        'Loss': loss,
        'Accuracy': accuracy,
        'Macro F1': macro_f1
    })

    print(f"\n{'='*40}")
    print("Loss:", loss)
    print(classification_report(y_test, predictions))

results_sgd_df = pd.DataFrame(sgd_results)

print("\nSGD Comparison:")
print(results_sgd_df.sort_values(
    by='Macro F1', ascending=False
))
```

```python

from sklearn.linear_model import SGDClassifier
from sklearn.model_selection import GridSearchCV

sgd = SGDClassifier(
    loss='log_loss',
    max_iter=2000,
    random_state=42,
    tol=1e-3
)

param_grid = {
    'alpha': [0.000001, 0.00001, 0.0001, 0.001, 0.01],
    'penalty': ['l1', 'l2', 'elasticnet'],
    'class_weight': ['balanced', None]
}

grid_search_sgd = GridSearchCV(
    estimator=sgd,
    param_grid=param_grid,
    cv=3,
    scoring='f1_macro',
    n_jobs=-1,
    verbose=1
)

grid_search_sgd.fit(X_train, y_train)

print("Best Parameters:")
print(grid_search_sgd.best_params_)

print("\nBest CV Macro F1:")
print(grid_search_sgd.best_score_)

best_sgd = grid_search_sgd.best_estimator_

y_pred_sgd_tuned = best_sgd.predict(X_test)

print("\nTest Accuracy:",
      accuracy_score(y_test, y_pred_sgd_tuned))

print("Test Macro F1:",
      f1_score(y_test, y_pred_sgd_tuned, average='macro'))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_sgd_tuned))
```
